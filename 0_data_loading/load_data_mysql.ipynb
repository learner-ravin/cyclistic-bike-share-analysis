#### This Python file load the csv files data into MySQL, This method is fast and automatic than the traditional MySQL import wizard.

```python
import pandas as pd
import glob
import os
```

### 1) Loading Datasets

```python
# Importing multiple files 
path = r"your_file_path\datasets\*.csv"            # your file path 
files = glob.glob(path)
```

```python
for file in files:
    filename = os.path.basename(file)
    # print(filename)
```

```python
# Creating dataframes from the csv files and storing in a list
dataframes = []

for file in files:
    df = pd.read_csv(file)
    dataframes.append(df)
```

### 2) Creating the new database

```python
import mysql.connector
```

```python
host = "your_host"                   #  hostname
password = "*************"           # your password
user = "your_username"               # user name
```

```python
database = mysql.connector.connect(
    host = host,
    password = password,
    user = user)

mycursor = database.cursor()
mycursor.execute("CREATE DATABASE IF NOT EXISTS cyclistic")
```

### 3) Loading data into the mysql database 

```python
from sqlalchemy import create_engine
```

```python
link = f"mysql+pymysql://{user}:{password}@{host}/cyclistic"
```

```python
engine = create_engine(link)
```

```python
df_names = ["jan", "feb", "march", "apr", "may", "june", "july", "aug", "sept", "oct", "nov", "dec"]
```

```python
# Loading these files to mysql

for name, df in zip(df_names, dataframes):
    with engine.begin() as conn:
        df.to_sql(
            name = name,
            con = conn,
            if_exists="replace",
            index=False)
```
